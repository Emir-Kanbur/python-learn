# NumPy

NumPy is a library for the Python, adding support for large, multi-dimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays.

Important things to cover:

* shape and reshape
* arraywise operations
* universal functions (pi, sqrt, ...)
* array slicing
* min max
* arange, linspace, zeros, ones
* sum
* random
* meshgrid

# Matplotlib

```python
%matplotlib inline
```

Matplotlib is a plotting library for Python and its numerical mathematics extension NumPy.

```python
import matplotlib as mpl
import matplotlib.pyplot as plt
```

## 2D

```python
N = 250
x = np.linspace(0, 2*np.pi*5, N)
y = np.cos(x) * np.exp(-x / 5)

plt.figure(figsize=(8, 4))
plt.plot(x, y, 'r')
plt.xlabel('x')
plt.ylabel('cos(x)*exp(-x/5)')
plt.show()
```

```python
N = 500
x = np.linspace(-2*np.pi*5, 2*np.pi*5, N)
y = np.linspace(-2*np.pi*5, 2*np.pi*5, N)

X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.cos(R) * np.exp(-R/5)

plt.figure(figsize=(8, 8))
plt.contour(X, Y, Z, 100);
```

```python
N = 500
x = np.linspace(-2*np.pi*5, 2*np.pi*5, N)
y = np.linspace(-2*np.pi*5, 2*np.pi*5, N)

X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.cos(R) * np.exp(-R/5)

plt.figure(figsize=(8, 8))
plt.contourf(X, Y, Z);
```

```python
N = 1000
x = np.linspace(-2*np.pi*5, 2*np.pi*5, N)
y = np.linspace(-2*np.pi*5, 2*np.pi*5, N)

X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)
Z = np.cos(R) * np.exp(-R/5)

plt.figure(figsize=(8, 8))
plt.pcolormesh(X, Y, Z, cmap='jet');
```

## 3D

```python
from mpl_toolkits import mplot3d
```

```python
plt.figure(figsize=(8, 8))
ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z, cmap='jet');
```

```python
N = 1000
x = np.linspace(-2*np.pi*5, 2*np.pi*5, N)
y = np.linspace(-2*np.pi*5, 2*np.pi*5, N)

X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos

plt.figure(figsize=(8, 8))
ax = plt.axes(projection='3d')
ax.plot_surface(X, Y, Z, cmap='jet');
```
