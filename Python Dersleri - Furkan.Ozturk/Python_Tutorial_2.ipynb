# <a id="loops">Loops</a>

**Loops**: You can loop over the elements of a list like this:

```python
animals = ['cat', 'dog', 'monkey']
animals[0]
```

```python
animals[2]
```

```python
animals = ['cat', 'dog', 'monkey', 2, 10, -6]
for x in animals:
    print(x)
# Prints "cat", "dog", "monkey", each on its own line.
```

```python
liste1 = [2, 5, 6, 12]
liste2 = [3, 8, 4, 11]

for n1 in liste1:
    for n2 in liste2:
        print(n1 + n2)
```

If you want access to the index of each element within the body of a loop, use the built-in `enumerate` function:

```python
animals = ['cat', 'dog', 'monkey', 2, 10, -6]
for idx, x in enumerate(animals):
    print('#%d: %s' % (idx + 1, x))
# Prints "#1: cat", "#2: dog", "#3: monkey", each on its own line
```

# <a id="Conditionals">Conditionals</a>

In a Python program, the if statement is how you perform a sort of decision-making. It allows for conditional execution of a statement or group of statements based on the value of an expression.

## if, elif, else

```python
for x in [-7, 0, 3, 13, 2.5, -4]:
    
    if x > 0:
        print(x, 'is ' + 'positive')
    elif x < 0:
        print(x, 'is ' + 'negative')
#     else:
#         print(x, 'is ' + 'zero')
```

## pass

```python
for x in [-7, 0, 3]:
    
    if x > 0:
        pass
    elif x < 0:
        print(x, 'is ' + 'negative')
    else:
        print(x, 'is ' + 'zero')
```

## while

```python
ind = 0;

while ind < 5:
    x = ind ** 2
    print(x)
    ind = ind + 1
```

# <a id="functions">Functions</a>
Python functions are defined using the def keyword. For example:

```python
def sign(x):
    if x > 0:
        return 'positive'
    elif x < 0:
        return 'negative'
    else:
        return 'zero'
```

```python
sign(0)
```

```python
for x in [-1, 0, 1]:
    print(sign(x))
# Prints "negative", "zero", "positive"
```

```python
def kare(x1, x2):
    y = x1 ** x2 + 5
    return y
```

```python
kare(2, 3)
```

```python
def func(x1, x2, kare=True):
    if kare:
        return x1**2 + x2**2
    else:
        return x1*x2
```

```python
func(2,3,kare=False)
```

We will often define functions to take optional keyword arguments, like this:

```python
def hello(name, loud=False):
    if loud:
        print('HELLO, %s!' % name.upper())
    else:
        print('Hello, %s' % name)

hello('Bob') # Prints "Hello, Bob"
hello('Fred', loud=True)  # Prints "HELLO, FRED!"
```

```python
func = lambda x: x**2 - 1
func(5)
```

```python
def func(x):
    y = x**2 - 1
    return y
```
